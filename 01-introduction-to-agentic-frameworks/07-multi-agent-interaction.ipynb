# Create Multi Agent

## 1. Import necessary packages

```python
from multi_agent_helpers import Agent, Tool, add, subtract, multiply
```

## 2. Create agent to add numbers

```python
add_agent = Agent(
    name="add_agent",
    role="Do the addition of two numbers",
    instructions="Help your peers with addition problems. Only use add tool.",
    tools=[Tool(add)],
)
```

## 3. Create agent to subtract numbers

```python
subtract_agent = Agent(
    name="subtract_agent",
    role="Do the subtraction of two numbers",
    instructions="Help your peers with subtraction problems. Only use subtract tool.",
    tools=[Tool(subtract)],
)
```

## 4. Create agent to multiple numbers

```python
multiply_agent = Agent(
    name="multiply_agent",
    role="Do the multiplication of two numbers",
    instructions="Help your peers with multiplication problems. Only use multiply tool.",
    tools=[Tool(multiply)],
)
```

## 5. Create main agent

```python
agent = Agent(
    "assistant", 
    peer_agents=[add_agent, subtract_agent, multiply_agent]
)
```

```python
agent.memory.reset()
# agent.peer_agents_map['add_agent'].memory.reset()
# agent.peer_agents_map['subtract_agent'].memory.reset()
# agent.peer_agents_map['multiply_agent'].memory.reset()
```

```python
agent.invoke("What is 56 multiplied by 12 then added to value of 67 subtracted by 34? Only use the available agents (add_agent, subtract_agent, multiply_agent). Do not use any other tools or calculators.")
```
