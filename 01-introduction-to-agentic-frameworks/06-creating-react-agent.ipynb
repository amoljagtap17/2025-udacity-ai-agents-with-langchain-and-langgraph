# Create ReAct Agent

## 1. Import necessary packages

```python
from react_agent_helpers import Agent, Tool, add, subtract, multiply
```

```python
funcs = [add, subtract, multiply]

tools = [Tool(func) for func in funcs]
```

```python
agent = Agent(
    tools=tools
)
```

```python
agent.memory.reset()
```

```python
agent.invoke("What is 45 multiplied by 32 then added to 23 subtracted by 12? Only use available tools appropriately. Don't perform any calculations yourself using calculators.")
```

```python
agent.memory.get_messages()
```
