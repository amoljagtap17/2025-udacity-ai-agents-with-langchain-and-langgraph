## 4. Define tools

```python
tools = [
    {
        "type": "function",
        "function": {
            "name": "multiply",
            "description": "Provides the result for multiplying two numbers",
            "parameters": {
                "type": "object",
                "properties": {
                    "num1": {
                        "type": "number",
                        "description": "The first number to multiply",
                    },
                    "num2": {
                        "type": "number",
                        "description": "The second number to multiply",
                    },
                },
                "required": ["num1", "num2"],
                "additionalProperties": False
            },
            "strict": True
        }
    }
]
```

```python
memory = Memory()
memory.add_message(role="system", content="You're a helpful assistant.")
```

```python
memory.get_messages()
```

```python
ai_message = chat_with_tools(
    "what is 32 multiplied by 7?",
    model=model,
    tools=tools,
    memory=memory,
)
```

```python
memory.get_messages()
```

```python
memory.last_message()['tool_calls']
```

```python
tool_call_id = memory.last_message()['tool_calls'][0].id

tool_call_id
```

```python
args = json.loads(memory.last_message()['tool_calls'][0].function.arguments)

args
```

```python
result = multiply(args["num1"], args["num2"])

result
```

```python
memory.add_message(role="tool", content=str(result), tool_call_id=tool_call_id)
```

```python
memory.get_messages()
```

```python
ai_message = chat_with_tools(
    model=model,
    tools=tools,
    memory=memory,
)
```
