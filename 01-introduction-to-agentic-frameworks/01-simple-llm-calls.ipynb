## 3. Import necessary packages

## 4. Instantiate OpenAI client

```python
load_dotenv()

client = OpenAI()
```

## 5. Define Parameters

```python
model = "gpt-5-nano-2025-08-07"

temperature = 1.0
```

# 6. Define System Prompt

```python
system_prompt = """You are a Senior Content Creator for a Social Networking platform.
Your job is to create engaging and relevant content for the platform's users.
You have a deep understanding of social media trends, user behavior, and content strategies."""
```

# 7. Create Reusble function for LLM calls

```python
def create_content(query:str,
                   client:OpenAI, 
                   system_prompt:str,
                   model:str,
                   temperature:float)->str:
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": query},
    ]
    
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature,
    )

    content = response.choices[0].message.content

    return content
```

## 8. Call create_content function

```python
# Senior React Developer input

react_developer_query = "Create a short intro for a Senior React Developer with 15 years of experience in building scalable web applications."
```

```python
content = create_content(
    query=react_developer_query,
    client=client,
    system_prompt=system_prompt,
    model=model,
    temperature=temperature
)
```
