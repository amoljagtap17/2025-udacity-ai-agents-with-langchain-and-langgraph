# Agent from Scratch

## 1. Import necessary packages

```python
from openai import OpenAI
from dotenv import load_dotenv
```

## 2. Instantiate OpenAI client

```python
load_dotenv()

client = OpenAI()
```

## 3. Define Parameters

```python
model = "gpt-5-nano-2025-08-07"

temperature = 1.0
```

## 4. Define System and User prompts

```python
system_prompt = "Act as Senior React Developer. You don't know anything about other programming language, so don't provide answers about languages like Java."
user_prompt = "What is the Java Virtual Machine?"
```

```python
messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt},
    ]

response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature,
    )

response.choices[0].message.content
```

## 5. Define Agent class

```python
class Agent():
    """A simple AI Agent"""

    def __init__(
        self,
        name:str = "Agent", 
        role:str = "Personal Assistant",
        instructions:str = "Assist the user with their queries",
        model:str = "gpt-5-nano-2025-08-07",
        temperature:float = 1.0,
    ):
        self.name = name
        self.role = role
        self.instructions = instructions
        self.model = model
        self.temperature = temperature

        self.client = OpenAI()
        
    def invoke(self, message: str) -> str:
        response = self.client.chat.completions.create(
            model = self.model,
            temperature=self.temperature,
            messages = [
                {
                    "role": "system",
                    "content": f"You're an helpful AI Agent, your role is {self.role}, " 
                               f"and you need to {self.instructions}",
                },
                {
                    "role": "user",
                    "content": message,
                }

            ]
        )
        return response.choices[0].message.content
```

## 6. Create Agent

```python
agent = Agent()
response = agent.invoke("What is React.js?")

print("Agent role :: ", agent.role)
print("Default Agent Response :: ", response)
```

```python
react_agent = Agent(
    role="React Developer", 
    instructions="Help users with React.js questions", 
    temperature=1.0
)

react_response = react_agent.invoke("What is the Virtual DOM?")

print("Agent role :: ", react_agent.role)
print("React Agent Response :: ", react_response)
```

```python
python_agent = Agent(
    role="Python Developer", 
    instructions="Help students solve Python programming problems."
)

python_response = python_agent.invoke("How to read a file in Python?")

print("Agent role :: ", python_agent.role)
print("Python Developer Response :: ", python_response)
```

```python
content_creator_agent = Agent(
    role="Content Creator", 
    instructions="Provide short bios for social media profiles.", 
    temperature=1.0
)

content_response = content_creator_agent.invoke("Create a short bio for a React Developer with 15 years experience.")

print("Agent role :: ", content_creator_agent.role)
print("Creative Agent Response :: ", content_response)
```
